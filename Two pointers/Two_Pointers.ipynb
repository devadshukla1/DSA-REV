## A. Opposite Direction Two-Pointer Technique (Target Sum / Two-Sum)

```python
def finderNum(numbers,target):
  left=0
  right=len(numbers)-1

  while left<right:

    total = numbers[left] + numbers[right]

    if (total==target):
      return(numbers[left], numbers[right])
    elif total>target:
      right -= 1
    else:
      left+=1
  return []

numbers=[1,2,3,4,5,6]
print(finderNum(numbers,9))
```

## B. Same Direction Two-Pointer Technique (Removing Duplicates in-Place)

```python
def clean(names):
  left=0

  for right in range(1,len(names)):
    if names[left]!=names[right]:
      left+=1
      names[left]=names[right]
  return left +1
names=["Aman","Aman", "Bhavna", "Chirag", "Chirag" ,"Diya"]
count=clean(names)
print(names[:count])
```
