# Complexity Analysis

#### 1- O(1)-Constant Time,workload remains unchanged regardless of data size

```python
marks=[88,90,72,7,92]
first_Marks=marks[0]
print(first_Marks) ## No matter if the list size get bigger the output remain
# constant and at same speed
```

#### 2. $O(\log n)$: Rises very gradually as data grows

```python
def find_Names(names,target):
  left=0
  right=len(names)-1

  while left<=right:
    middle=(left+right)//2

    if names[middle]==target:
      return middle

    elif names[middle]<target:
      left= middle+1

    else:
      right= middle-1

  return -1


names=["Aayush", "Bhavna","Chirag" ,"Diya" ,"sneha","uday","zara"]
target="sneha"

result=find_Names(names,target)

if result!=-1:
  print("Founded at index",result)
else:
  print("Not Founded")
```

#### 3. $O(n)$: Workload grows steadily in direct proportion to data size1011

```python
def find_Ppr(papers,target):

  for i in papers:
    if i==target:
      return True

  return False


names=["Aayush", "Bhavna","Chirag" ,"Diya" ,"sneha","uday","zara"]
target="Diya"

result=find_Ppr(names,target)

if result:
  print("Found",result)
else:
  print("Not found")
```

#### 4.$O(n^2)$- Rises steeply; workload increases sharply with data growth

```python
def find_Numbers(numbers):
  for i in range(len(numbers)):
    for j in range(len(numbers)):
      if i!=j and numbers[i]==numbers[j]:
        return True
  return False


numbers=[10,22,35,22,48]
result= find_Numbers(numbers)

if result:
  print("Found duplicates")
else:
  print("Not found")
```

# TIME AND SPACE COMPLEXITY

#### WAY 1 (LOW MEMORY,LOW SPACE)

```python
def findDuplicates_Slow(numbers):
  for i in range(len(numbers)):
    for j in range(len(numbers)):
      if i!=j and numbers[i]==numbers[j]:
        return True
  return False


numbers=[10,20,30,40,50,10]

result=findDuplicates_Slow(numbers)

if result:
  print("Found slow")
else:
  print("not found")
```

WAY 2 (HIGH MEMORY,HIGH SPACE)

```python
def findDuplicates_Fast(numbers):
  seen=set()

  for i in numbers:
    if i in seen:
      return True
    seen.add(i)
  return False

numbers=[10,20,30,40,50,10]

result=findDuplicates_Fast(numbers)

if result:
  print("Found fast")
else:
  print("not found")
```
