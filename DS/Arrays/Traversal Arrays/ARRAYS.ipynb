# Operations on Array

## 1. Traversal in Arrays

### A. Liner Traversal

```python
arr=[1,2,3,4,5]

print("Linear Traversal: ",end=" ")
for i in range(len(arr)):
  print(arr[i],end=" ")

#Time complexity:  O(n)
```

### B. Reverse Traversal

```python
arr=[1,2,3,4,5]

print("Reverse Traversal: ",end=" ")
for i in range(len(arr)-1,-1,-1): #range(start[5-1:4 i.e start from 4th index], stop[prints till 0th index -1 excluded], step)
  print(arr[i],end=" ")
```
