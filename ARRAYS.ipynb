# Operations on Array

## 1. Traversal in Arrays

### A. Liner Traversal

```python
arr=[1,2,3,4,5]

print("Linear Traversal: ",end=" ")
for i in range(len(arr)):
  print(arr[i],end=" ")

#Time complexity:  O(n)
```

### B. Reverse Traversal

```python
arr=[1,2,3,4,5]

print("Reverse Traversal: ",end=" ")
for i in range(len(arr)-1,-1,-1): #range(start[5-1:4 i.e start from 4th index], stop[prints till 0th index -1 excluded], step)
  print(arr[i],end=" ")
```

# 3.Insertion

## A. At beginning

```python
def insertionAtBeggining(numbers,element):
  numbers.insert(0,element)
  for i in range(len(numbers)):
    print(arr[i], end=" ")

arr=[1,2,3,4,5]
insertionAtBeggining(arr,10)
```

## B. Insert Element at a Given Position in an Array

```python
def insertAtPos(arr,element,pos):
  arr.insert(pos,element)
  for i in range(len(arr)):
    print(arr[i], end=" ")

arr=[1,2,3,4,5]
insertAtPos(arr,10,3)
```

## C. Insert at the end

```python
def InsertAtEnd(arr,element):
  arr.append(element)
  for i in range(len(arr)):
    print(arr[i],end=" ")

arr=[1,2,3,4,5]
InsertAtEnd(arr,100)
```

# 4. Deletion

## A.Delete an Element from the Beginning of an Array

```python
def ElementDeletion(arr,indexValue):
  if 0<=indexValue < len(arr):
      del arr[indexValue]
      print("Deleted element from index",indexValue)
      return arr
  else:
    print("Element is not in an array")

arr=[1,2,3,4,5]
ElementDeletion(arr,0)
```

## B.Delete an Element from a Given Position in an Array

```python
def ElementDeletionPos(arr,indexValue):
  for i in range(len[arr]):
    if i==indexValue:
      del arr[indexValue]
      print("Item deleted from specific index",indexValue)
      return arr
  else:
    print("Not deleted")

arr=[1,2,3,3,4,5]
ElementDeletion(arr,3)
```

## C.Delete First Occurrence of Given Element from an Array

```python
def removeElement(arr,element):
  for i in range(len(arr)):
    if arr[i]==element:
        arr.remove(element)
        print("ELEMENT DELETED")
        return arr
  else:
    print("Not deleted")

arr=[1,2,3,4,5]
removeElement(arr,4)
```

## D.Remove All Occurrences of an Element in an Array

```python
def removeAllElement(arr,element):
  i=0
  while i<len(arr):
    if arr[i]==element:
        arr.remove(element)
    else:
      i+=1
  return arr


arr=[1,2,3,3,4,5]
removeAllElement(arr,3)
```

## E.Remove last element
