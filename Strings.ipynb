# 1. Length of strings

```python
def length_Finder(s):
    return len(s)

text="Hello worlds"
print(length_Finder(text))
```

# 2. Are text same

```python
def checker(s1, s2):

    if s1 == s2:
        return "Both are same"
    else:
        return "Both are different"


text = "Hello worlds"
text2 = "Hello worlds"

print(checker(text, text2))
```

# 3. Search for the characters

```python
def finder(s,ch):
  cnt=0
  for i in range(len(s)):
    if s[i]==ch:
      cnt+=1
  return f"Found {cnt} times"


s = "geeksforgeeks"
ch = 'k'
print(finder(s,ch))
```

# 4. Concatenating texts

```python
def conncecter(s1,s2):
  return s1 + s2

s1="Devad"
s2=" Shukla"

print(conncecter(s1,s2))
```

# 5 Reverse the string

```python
def reverser(s1):
  res = []
  for i in range(len(s1)-1,-1,-1):
    res.append(s1[i])

  return " ".join(res) #convert list to string

s1="Devad"
print(reverser(s1))
```

# 6. Insert a character in String at a Given Position

```python
def inserter(s,ch,pos):
  return s[:pos]+ ch+s[pos:]

s = "Geeks"
print(inserter(s, 'A', 3))
```

# 8. Remove all occurance of word in given string

```python
# Using built-in function

def remover(s,ch):
  return s.replace(ch," ")

s1="Devad"
print(remover(s1,"D"))
```

```python
# Creating our function

def remover_ch(s,ch):
  j=0
  s_list=list(s)

  for i in range(len(s_list)):
    if s_list[i]!=ch:
      s_list[j]=s_list[i]
      j+=1

  return ''.join(s_list[:j])

s = "geeksforgeeks"
s = remover_ch(s, 'g')
print(s)

```
